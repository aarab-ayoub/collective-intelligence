# 02 — Evaluate Baseline
Loads baseline metrics and prints summary.

```python
import json
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path('/Users/ayoub/work/MS-DS_ML_Projects/IOT/collective-intelligence')
METRICS = PROJECT_ROOT / 'results' / 'baseline' / 'baseline_metrics.json'
if not METRICS.exists():
    raise FileNotFoundError(f'Missing metrics file: {METRICS}')

with open(METRICS) as f:
    m = json.load(f)

print('=== BASELINE TEST RESULTS ===')
print('Accuracy          :', f"{100*m['test_accuracy']:.2f}%")
print('Macro F1          :', f"{m['test_macro_f1']:.4f}")
print('Per-class F1      :', m['per_class_f1'])
print('Model size        :', f"{m['model_size_mb']:.2f} MB")
print('Avg inference time:', f"{m['avg_inference_time_ms']:.2f} ms")

ok = (m['test_accuracy'] > 0.85) and (m['test_macro_f1'] > 0.75) and (m['model_size_mb'] < 50) and (m['avg_inference_time_ms'] < 100)
print('✅ BASELINE CRITERIA MET' if ok else '❌ CRITERIA NOT MET')

if 'history' in m and m['history']:
    display(pd.DataFrame(m['history']).tail())
```
